```python
from openai import OpenAI
from bs4 import BeautifulSoup
import requests
```

```python
ollama_baseurl = "http://localhost:11434/v1"

ollama = OpenAI(base_url=ollama_baseurl,api_key="ollama")

def ollam_funct(prompt, model):
    response = ollama.chat.completions.create(
        model = model,
        messages=[{"role":"user","content":prompt}]
        
    )
    return response.choices[0].message.content

# prompt = "Tell me one fun fact"
# model="gemma3:270m"
# ollam_funct(prompt,model)
```

```python
url = "https://docs.ollama.com/cloud"
headers = {
    "User-Agent": "Chrome/136.0.0.0"
}

response = requests.get(url, headers=headers)

if response.status_code == 200:
    print("Successfully Fetched")
    soup = BeautifulSoup(response.text, "html.parser")

    for tags in soup(["script", "style", "nav", "footer"]):
        tags.decompose()

    for p in soup.find_all('span',attrs={"data-as":"p"}):
        if p.find('a'):
            p.decompose()
    text = soup.get_text(separator="",strip=True)
    #adding limit to the text
    text= text[:10000]
else:
    print("Failed to fetch")
```

```python
prompt = f"""
Summarize the following text into clear bullet points.

Rules:
- Use simple language
- Avoid escape characters
- Use proper markdown formatting
- Keep the summary concise

Text:
{text}

"""
model = "llama3.2"
ollam_funct(prompt,model)
```
