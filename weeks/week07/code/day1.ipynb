```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display
```

```python
load_dotenv()
geminiAPIKey = os.getenv('GEMINI_API_KEY')
groqApiKey = os.getenv('GROQ_API_KEY')
```

```python
groq_url = "https://api.groq.com/openai/v1"
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"
```

```python
gemini = OpenAI(api_key=geminiAPIKey, base_url=gemini_url)
groq = OpenAI(api_key=groqApiKey,base_url=groq_url)
```

```python
tell_a_fact = [
    {"role": "user", "content": "Tell a me facts about LLM Engineering"}
]

response = groq.chat.completions.create(model="gemini-2.5-pro", messages=tell_a_fact)
print(response)
```
