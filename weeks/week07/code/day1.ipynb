```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display
```

```python
load_dotenv()
geminiAPIKey = os.getenv('GEMINI_API_KEY')
groqApiKey = os.getenv('GROQ_API_KEY')
```

```python
groq_url = "https://api.groq.com/openai/v1"
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"
llama_url  = "http://localhost:11434/v1"
```

```python
gemini = OpenAI(api_key=geminiAPIKey, base_url=gemini_url)
groq = OpenAI(api_key=groqApiKey,base_url=groq_url)
llama = OpenAI(api_key='apiKey',base_url=llama_url)
```

```python
tell_a_fact = [
    {"role": "user", "content": "Tell a me facts about LLM Engineering"}
]
```

```python
def differentModelTesting(llm,models,message):
   response = ""
   stream = llm.chat.completions.create(model=models,messages= message,stream=True)
   for chunk in stream:
       response += chunk.choices[0].delta.content or ''
       # clear_output(wait=True)
      
   display(Markdown(response))
```

```python
differentModelTesting(llm=gemini,models="models/gemini-2.5-flash",message=tell_a_fact)
```

```python
differentModelTesting(llm=groq,models= "openai/gpt-oss-20b",message=tell_a_fact)
```

```python
llamaModel = "llama3.2:latest"
differentModelTesting(llm=llama,models=llamaModel,message=tell_a_fact)
```
